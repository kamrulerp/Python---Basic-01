**3. String Reverse:** Write a Python function to reverse a given string and return the reversed string.

```python
def reverse_string(string) :
    
    r = len(string) -1
    reverse = ''
    for i in range(r, -1, -1) :
       reverse += string[i]
    
    return reverse

x = 'kamrul'
print(reverse_string(x))
```

**6. Data Type Checker:** Write a Python function that takes a variable as input and returns the data type of the variable as a string (e.g., “int”, “float”, “str”, “list”, etc.).

```python
def data_type_checker(variable):
    if isinstance(variable, int):
        return "int"
    elif isinstance(variable, float):
        return "float"
    elif isinstance(variable, str):
        return "str"
    elif isinstance(variable, list):
        return "list"
    elif isinstance(variable, dict):
        return "dict"
    elif isinstance(variable, tuple):
        return "tuple"
    elif isinstance(variable, set):
        return "set"
    elif isinstance(variable, bool):
        return "bool"
    elif variable is None:
        return "NoneType"
    else:
        return ""


x = input('Input a variable: ')
eval_input = eval(x)
print(data_type_checker(eval_input))
```

**7. String Palindrome:** Write a Python function to check if a given string is a palindrome or not.

```python
def isPalindrome(string) :
    l = 0
    r = len(string) - 1
    while l < r :
        if string[l] != string[r] :
            return False
        l += 1
        r -= 1
    return True

x = "robin"
print(isPalindrome(x))
```

**18. Quadratic Equation Solver:** Write a Python program that takes the coefficients (a, b, c) of a quadratic equation as input and calculates and prints the real roots (if they exist) or a message indicating the complex roots.

ax2+bx+c=0

```python
import cmath

a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

discriminant = b**2 - 4*a*c

root1 = (-b + cmath.sqrt(discriminant)) / (2 * a)
root2 = (-b - cmath.sqrt(discriminant)) / (2 * a)

if discriminant > 0:
    print(f"The equation has two distinct real roots: {root1.real:.2f} and {root2.real:.2f}")
elif discriminant == 0:
    print(f"The equation has one real root: {root1.real:.2f}")
else:
    print(f"The equation has complex roots: {root1} and {root2}")
```

**27. Prime Number Checker:** Write a Python program using a while loop to check if a given number N is prime or not.

```python
def check_prime(n):    
    if n <= 1:
        print(f"{n} is not prime number")
        return
    
    if n <= 3:
        print(f"{n} is prime number")
        return
    
    if n % 2 == 0:
        print(f"{n} is not prime number")
        return
    
    divisor = 3
    
    while divisor <= n:
        if n % divisor == 0:
            print(f"{n} is not prime number (divisible by {divisor})")
            return
        divisor += 2
    
    print(f"{n} is prime number")

n = int(input("Enter a number: "))
check_prime(n)
```

**28. List Manipulation:** Given a list of integers, write a Python program using a for loop to find the sum, average, maximum, and minimum values in the list. <br>
List: [10, -5, 0, 15, -10] <br>
Sum: 10 <br>
Average: 2.00 <br>
Maximum: 15 <br>
Minimum: -10

```python
lists = [10, -5, 0, 15, -10]

total_sum = 0
maximum = lists[0]
minimum = lists[0]

for num in lists:
    total_sum += num
    
    if num > maximum:
        maximum = num
        
    if num < minimum:
        minimum = num
    
average = total_sum / len(lists)

print("Sum: ",total_sum)
print("Average: ",average)
print("Maximum: ",maximum)
print("Minimum: ",minimum)
```
