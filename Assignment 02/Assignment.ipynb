```python
import pandas as pd
import numpy as np
```

# Given Data set

```python
df = pd.read_csv('vegitable.csv') 
df.head() 
```

# Task 1: Calculate slope (M) and intercept (C) using Ordinary Least Squares

```python
x = df['Weight']
y = df['Price']

mean_x = np.mean(x)
mean_y = np.mean(y)
```

```python
print(mean_x)
print(mean_y)
```

```python
# Calculating slope (M)
dev_x = x - mean_x
dev_y = y - mean_y

m = np.sum(dev_x * dev_y) / np.sum(dev_x ** 2)
# y = mx + c

print("m = ",m)
```

```python
# Calculating intercept (C)
c = mean_y - m * mean_x
print("c = ",c)
```

```python
# Predict price when weight = 6.50
weight_pred = 6.50
price_pred = m * weight_pred + c

print(price_pred)
```

```python
print(f"Slope (m) = {m:.6f}")
print(f"Intercept (c) = {c:.6f}")
print(f"Regression equation: y = {m:.4f}x + {c:.4f}")
print()
```

```python
df['Predicted_Price'] = m * df['Weight'] + c
```

```python
df['Residuals'] = df['Price'] - df['Predicted_Price']
```

```python
print(df.round(4))
```

```python
# calculate L1 loss
L1_loss = df['Residuals'].abs().sum()
L1_loss
```

```python
# Calculate l2 loss
L2_loss = (df['Residuals']**2).sum()
L2_loss
```

```python
n = len(df)
MSE = L2_loss / n
MAE = L1_loss / n

print(f"MSE = {MSE:.2f}, MAE = {MAE:.2f}")
```
